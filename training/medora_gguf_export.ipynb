```python
from unsloth import FastModel

adapter_repo = "Wallfou/medora-gemma-4-e2b-lora"

model, tokenizer = FastModel.from_pretrained(
    model_name=adapter_repo,
    max_seq_length=2048,
    load_in_4bit=True,
)

# Save merged 16-bit HF model. We do GGUF conversion ourselves in the next cell
# because Unsloth's wrapper is currently broken against latest llama.cpp.
model.save_pretrained_merged(
    "medora-gemma4-merged",
    tokenizer,
    save_method="merged_16bit",
)
print("Merge done. HF model at medora-gemma4-merged/")
```

```python
!git clone --depth 1 https://github.com/ggml-org/llama.cpp.git
!pip install -r llama.cpp/requirements.txt

!cd llama.cpp && python convert_hf_to_gguf.py \
    ../medora-gemma4-merged \
    --outfile ../medora-gemma4.BF16.gguf \
    --outtype bf16
```

```python
!cd llama.cpp && cmake -B build -DGGML_CUDA=OFF -DLLAMA_CURL=OFF && \
    cmake --build build --config Release --target llama-quantize -j

!./llama.cpp/build/bin/llama-quantize \
    medora-gemma4.BF16.gguf \
    medora-gemma4.Q4_K_M.gguf \
    Q4_K_M
```

```python
import os

modelfile = """FROM ./medora-gemma4.Q4_K_M.gguf

TEMPLATE \"\"\"{{ if .System }}<start_of_turn>user
{{ .System }}<end_of_turn>
{{ end }}{{ range .Messages }}<start_of_turn>{{ if eq .Role \"assistant\" }}model{{ else }}user{{ end }}
{{ .Content }}<end_of_turn>
{{ end }}<start_of_turn>model
\"\"\"

PARAMETER stop \"<end_of_turn>\"
PARAMETER temperature 0.2
"""

os.makedirs("medora-gemma4-gguf_gguf", exist_ok=True)
!cp medora-gemma4.Q4_K_M.gguf medora-gemma4-gguf_gguf/

with open("medora-gemma4-gguf_gguf/Modelfile", "w") as f:
    f.write(modelfile)

!zip -r medora-gemma4-ollama.zip medora-gemma4-gguf_gguf

from google.colab import files
files.download("medora-gemma4-ollama.zip")
```
