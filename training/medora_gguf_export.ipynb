```python
from unsloth import FastModel

adapter_repo = "Wallfou/medora-gemma-4-lora"

model, tokenizer = FastModel.from_pretrained(
    model_name=adapter_repo,
    max_seq_length=2048,
    load_in_4bit=True,
)

model.save_pretrained_gguf(
    "medora-gemma4-gguf",
    tokenizer,
    quantization_method="q4_k_m",
)
```
